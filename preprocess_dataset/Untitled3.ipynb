```python
import numpy as np

arr = np.array([[1, 2, 2, 3, 3, 3, 4]])
print(arr.shape[1])

def get_max_item_freq(np_arr):
    # Sort the array
    sorted_arr = np.sort(np_arr)
    # Find where the value changes
    changes = np.diff(sorted_arr) != 0
    # Get run lengths of identical values
    run_lengths = np.diff(np.concatenate(([0], np.where(changes)[0] + 1, [len(arr)])))
    # Get max frequency
    max_frequency = run_lengths.max()
    return max_frequency
    
max_frequency = get_max_item_freq(arr)
print(max_frequency)
```
