```python
import json
def gen_instruction(data):
    middle = data['msa'][1]
    middle_start = middle['start']
    middle_end = middle['end']
    
    instruction = (
        "Transform the music from noisy to clean by generating music for the noisy middle segment. " 
        f"Middle segment starts at {middle_start} seconds and ends at {middle_end} seconds. " 
        "Keep beginning segment and end segment identical."
    )

    return instruction
```

```python
json_file_path = "/homes/al4624/Documents/YuE_finetune/YuE_finetune_trans_gen/preprocess_dataset/output.jsonl"
json_line = None
with open(json_file_path, 'r') as file:
    # Read the first line
    json_line = file.readline()
data = json.loads(json_line)
print(gen_instruction(data))
```
